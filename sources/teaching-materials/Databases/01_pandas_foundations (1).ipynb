## What is a DataFrame?

A DataFrame is pandas' primary data structure. You can think of it as a table, similar to a spreadsheet or a database table. Each column has a name and contains values of a particular type, and each row represents one observation or record.

The most intuitive way to create a DataFrame is from a Python dictionary. Each key in the dictionary becomes a column name, and the corresponding list becomes that column's values. Let's create a small dataset about renewable energy production in different countries.

```python
# Creating a DataFrame from a dictionary
# Each key is a column name, each value is a list of data for that column

energy_data = {
    'country': ['Iceland', 'Norway', 'Sweden', 'Denmark', 'Germany', 'France', 'Spain', 'Italy'],
    'renewable_percent': [85.0, 71.5, 56.9, 67.0, 46.3, 24.7, 47.3, 37.4],
    'solar_capacity_gw': [0.2, 0.3, 1.9, 2.4, 66.5, 15.7, 28.0, 25.1],
    'wind_capacity_gw': [0.0, 5.2, 14.0, 7.8, 66.3, 20.4, 29.8, 12.0],
    'hydro_capacity_gw': [2.1, 33.0, 16.5, 0.0, 5.6, 25.5, 20.4, 22.6],
    'population_millions': [0.4, 5.5, 10.5, 5.9, 84.0, 68.0, 47.5, 58.9]
}

df = pd.DataFrame(energy_data)
df
```

When you display a DataFrame in Jupyter, you see a nicely formatted table. Notice that pandas has automatically added an index on the left side, starting from 0. This index is how pandas identifies each row.

The data above represents approximate renewable energy statistics for European countries. `renewable_percent` shows the percentage of electricity from renewable sources, while the capacity columns show installed generation capacity in gigawatts. These figures are simplified for teaching purposes, but they reflect real patterns: Nordic countries tend to have high renewable percentages largely due to hydropower, while Germany and Spain have invested heavily in solar and wind.

## Exploring the Structure of a DataFrame

Before doing any analysis, it helps to understand what you're working with. Pandas provides several methods for inspecting a DataFrame's structure. Let's explore what each one tells us.

```python
# .shape tells us the dimensions: (rows, columns)
print("Shape of the DataFrame:")
print(df.shape)
print()

# This means we have 8 rows (countries) and 6 columns (variables)
```

```python
# .columns gives us the column names as an Index object
print("Column names:")
print(df.columns)
print()

# You can convert this to a regular Python list if needed
print("As a list:")
print(list(df.columns))
```

```python
# .dtypes shows the data type of each column
print("Data types:")
print(df.dtypes)
```

Understanding data types matters because different operations are available for different types. The `object` type typically means text/strings, while `float64` means decimal numbers. If you tried to calculate the mean of the `country` column, pandas would raise an error because averaging text makes no sense.

The `.info()` method combines several pieces of information into one summary.

```python
# .info() provides a comprehensive overview
df.info()
```

The info output tells us the DataFrame has 8 entries (rows), numbered 0 to 7. It lists each column with its count of non-null values and data type. The "non-null" count is important because real-world data often has missing values, and this helps you spot them quickly. The memory usage at the bottom becomes relevant when working with large datasets.

## Accessing Columns

There are two main ways to access a single column in pandas. Both return a Series, which is pandas' one-dimensional data structure (like a single column extracted from the table).

```python
# Method 1: Bracket notation with the column name as a string
df['country']
```

```python
# Method 2: Dot notation (only works if the column name has no spaces or special characters)
df.renewable_percent
```

The bracket notation is more reliable because it works with any column name, including those with spaces or that happen to match pandas method names. Dot notation is convenient for quick exploration but can cause confusion if your column is named something like `count` or `mean`.

To select multiple columns, pass a list of column names inside the brackets.

```python
# Selecting multiple columns returns a DataFrame (not a Series)
df[['country', 'renewable_percent', 'population_millions']]
```

## Accessing Rows

Pandas provides two main indexers for accessing rows: `.loc[]` and `.iloc[]`. The difference is fundamental and worth understanding clearly.

`.iloc[]` uses integer positions, like indices in a Python list. The first row is at position 0, the second at position 1, and so on. This is positional indexing.

`.loc[]` uses the actual index labels. In our current DataFrame, the index labels happen to be integers 0-7, but they don't have to be. If we set the country names as our index, `.loc[]` would use those instead.

```python
# .iloc uses integer position
# Get the row at position 0 (first row)
df.iloc[0]
```

```python
# Get rows at positions 2 through 4 (inclusive of 2, exclusive of 5)
df.iloc[2:5]
```

```python
# .loc uses index labels
# Currently our index labels are 0-7, so this looks similar
df.loc[0]
```

The distinction becomes clearer when we change the index. Let's set the country names as our index.

```python
# Set 'country' as the index
df_indexed = df.set_index('country')
df_indexed
```

```python
# Now .loc uses country names
df_indexed.loc['Norway']
```

```python
# But .iloc still uses integer positions
df_indexed.iloc[1]  # Still gets the second row (Norway)
```

```python
# You can select multiple rows with .loc using a list of labels
df_indexed.loc[['Iceland', 'Norway', 'Sweden']]
```

## Combining Row and Column Selection

Both `.loc` and `.iloc` accept two arguments: rows first, then columns, separated by a comma.

```python
# Get renewable_percent for Norway using the indexed DataFrame
df_indexed.loc['Norway', 'renewable_percent']
```

```python
# Get specific rows and columns
df_indexed.loc[['Iceland', 'Norway', 'Sweden'], ['renewable_percent', 'hydro_capacity_gw']]
```

```python
# The same selection using .iloc (positions instead of labels)
# Rows 0, 1, 2 and columns 0, 4
df_indexed.iloc[0:3, [0, 4]]
```

## Filtering Data with Conditions

One of the most common operations is filtering rows based on some condition. This uses boolean indexing, which works by creating a series of True/False values and using it to select rows.

```python
# First, let's see what a condition looks like
# This creates a Series of True/False values
df['renewable_percent'] > 50
```

```python
# When you put this condition inside brackets, pandas keeps only the True rows
df[df['renewable_percent'] > 50]
```

The logic here is worth pausing on. The expression `df['renewable_percent'] > 50` compares each value in the column to 50 and returns True or False. When you use this boolean Series inside `df[...]`, pandas includes rows where the value is True and excludes rows where it's False. This pattern is extremely common in data analysis.

```python
# Countries with significant solar capacity (more than 10 GW)
df[df['solar_capacity_gw'] > 10]
```

```python
# You can combine conditions using & (and) or | (or)
# Important: each condition must be in parentheses

# Countries with both high renewable percent AND large population
df[(df['renewable_percent'] > 40) & (df['population_millions'] > 10)]
```

```python
# Countries that are EITHER very high in renewables OR have very high wind capacity
df[(df['renewable_percent'] > 70) | (df['wind_capacity_gw'] > 60)]
```

## Basic Statistical Summaries

Pandas makes it easy to calculate summary statistics. The `.describe()` method gives you a comprehensive overview of all numeric columns.

```python
df.describe()
```

The output shows count (number of non-null values), mean, standard deviation, minimum, 25th percentile, median (50th percentile), 75th percentile, and maximum for each numeric column.

You can also calculate individual statistics on specific columns.

```python
print(f"Mean renewable percentage: {df['renewable_percent'].mean():.1f}%")
print(f"Median renewable percentage: {df['renewable_percent'].median():.1f}%")
print(f"Total solar capacity: {df['solar_capacity_gw'].sum():.1f} GW")
print(f"Maximum wind capacity: {df['wind_capacity_gw'].max():.1f} GW")
```

## Creating New Columns

You often need to calculate new variables from existing ones. This is straightforward: just assign to a new column name.

```python
# Calculate total renewable capacity (solar + wind + hydro)
df['total_renewable_gw'] = df['solar_capacity_gw'] + df['wind_capacity_gw'] + df['hydro_capacity_gw']
df
```

```python
# Calculate renewable capacity per million people
df['renewable_per_capita'] = df['total_renewable_gw'] / df['population_millions']
df
```

The per-capita calculation reveals something interesting: Norway's renewable capacity per million people is far higher than Germany's, even though Germany has more total capacity. This is the kind of insight that comes from deriving new variables.

## Sorting Data

The `.sort_values()` method arranges rows based on one or more columns.

```python
# Sort by renewable percentage, highest first
df.sort_values('renewable_percent', ascending=False)
```

```python
# Sort by total renewable capacity, lowest first
df.sort_values('total_renewable_gw', ascending=True)
```
