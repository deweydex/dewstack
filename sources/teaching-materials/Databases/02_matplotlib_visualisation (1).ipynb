# Visualising Data with Matplotlib

This notebook introduces matplotlib, Python's foundational plotting library. By the end, you will be able to create line charts, multi-line plots, figures with multiple subplots, histograms, and stacked histograms. The examples continue our environmental theme, using data about temperature changes, emissions, and energy production.

Matplotlib was created by John Hunter in 2003, inspired by MATLAB's plotting capabilities. The library's design reflects this heritage, which explains some of its conventions. While there are now many Python visualisation libraries (seaborn, plotly, altair, and others), matplotlib remains the foundation that most others build upon. Understanding matplotlib gives you both practical skills and the background to learn other libraries more easily.

The ultimate goal of data visualisation is communication. A good chart makes patterns visible that would be hidden in tables of numbers. As you work through these examples, consider not just how to create the plots but why particular choices help or hinder understanding.

## Setup

We import matplotlib's pyplot module with the conventional alias `plt`. The `%matplotlib inline` magic command tells Jupyter to display plots directly in the notebook rather than in separate windows.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

%matplotlib inline

# Set a clean style
plt.style.use('seaborn-v0_8-whitegrid')
```

## Creating Data for Visualisation

Let's create a dataset representing global temperature anomalies over time. Temperature anomaly means the difference from a baseline average, which is how climate scientists typically present warming trends. These values are simplified approximations of real data from NASA and NOAA.

```python
# Global temperature anomaly data (simplified, based on NASA GISS)
# Values represent degrees Celsius above/below a 1951-1980 baseline

years = list(range(1880, 2025, 5))

# These approximate the actual trend: relatively stable until ~1980, then accelerating warming
temp_anomaly = [
    -0.19, -0.27, -0.31, -0.24, -0.38, -0.26, -0.14, -0.31, -0.19, -0.07,
    -0.08, -0.01, 0.03, -0.10, -0.08, 0.07, 0.09, 0.03, 0.20, 0.26,
    0.34, 0.42, 0.54, 0.68, 0.75, 0.92, 1.02, 1.18, 1.25
]

# Create a DataFrame
temp_df = pd.DataFrame({
    'year': years,
    'anomaly': temp_anomaly
})

temp_df.head(10)
```

## A Simple Line Chart

The most basic plot uses `plt.plot()` with x and y values. Let's visualise the temperature trend.

```python
plt.plot(temp_df['year'], temp_df['anomaly'])
plt.show()
```

This works, but it's missing context. A viewer wouldn't know what the axes represent or what the chart is about. Let's add proper labels and a title.

```python
plt.figure(figsize=(10, 6))  # Make it larger for better readability

plt.plot(temp_df['year'], temp_df['anomaly'])

plt.xlabel('Year')
plt.ylabel('Temperature Anomaly (°C)')
plt.title('Global Temperature Anomaly, 1880-2020')

plt.show()
```

The `figsize` parameter sets the figure dimensions in inches as (width, height). Larger figures are generally easier to read, especially when you're going to present or share them.

Let's enhance this further with a horizontal reference line at zero and some colour to indicate warming versus cooling.

```python
plt.figure(figsize=(10, 6))

# Draw the baseline reference first so it appears behind the data
plt.axhline(y=0, color='gray', linestyle='--', linewidth=1, alpha=0.7)

# Plot the data with customised appearance
plt.plot(temp_df['year'], temp_df['anomaly'], 
         color='crimson', 
         linewidth=2,
         marker='o',
         markersize=4)

plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature Anomaly (°C)', fontsize=12)
plt.title('Global Temperature Anomaly Relative to 1951-1980 Average', fontsize=14)

# Add some annotation
plt.text(1885, 0.05, 'Baseline (0°C)', color='gray', fontsize=10)

plt.tight_layout()  # Prevents labels from being cut off
plt.show()
```

## Line Chart Customisation Options

Matplotlib offers extensive control over line appearance. Here are the key parameters for `plt.plot()`:

The `color` parameter accepts named colours ('red', 'blue', 'green'), hex codes ('#FF5733'), or RGB tuples ((0.5, 0.2, 0.8)). The `linewidth` (or `lw`) controls thickness. The `linestyle` (or `ls`) can be solid ('-'), dashed ('--'), dotted (':'), or dash-dot ('-.'). The `marker` parameter adds points at each data value, with options including 'o' (circle), 's' (square), '^' (triangle), and 'x'. The `alpha` parameter controls transparency from 0 (invisible) to 1 (opaque).

```python
# Demonstrating different line styles
plt.figure(figsize=(10, 6))

x = np.linspace(0, 10, 50)

plt.plot(x, np.sin(x), color='blue', linestyle='-', linewidth=2, label='Solid')
plt.plot(x, np.sin(x + 1), color='green', linestyle='--', linewidth=2, label='Dashed')
plt.plot(x, np.sin(x + 2), color='red', linestyle=':', linewidth=2, label='Dotted')
plt.plot(x, np.sin(x + 3), color='purple', linestyle='-.', linewidth=2, label='Dash-dot')

plt.legend()
plt.title('Line Style Options')
plt.show()
```

## Multiple Lines on One Chart

Comparing multiple data series on the same axes is common. Let's create a dataset showing energy production from different sources over time.

```python
# Energy production data (simplified global figures in exajoules)
energy_years = list(range(1990, 2025, 5))

energy_df = pd.DataFrame({
    'year': energy_years,
    'coal': [89, 92, 95, 120, 145, 160, 155],
    'oil': [135, 140, 152, 165, 175, 180, 178],
    'gas': [72, 80, 88, 100, 115, 130, 140],
    'nuclear': [20, 24, 26, 27, 26, 27, 28],
    'renewables': [5, 8, 12, 18, 30, 52, 78]
})

energy_df
```

```python
plt.figure(figsize=(12, 7))

# Plot each energy source
plt.plot(energy_df['year'], energy_df['coal'], label='Coal', color='#4a4a4a', linewidth=2, marker='o')
plt.plot(energy_df['year'], energy_df['oil'], label='Oil', color='#2d5a27', linewidth=2, marker='s')
plt.plot(energy_df['year'], energy_df['gas'], label='Natural Gas', color='#1a75ff', linewidth=2, marker='^')
plt.plot(energy_df['year'], energy_df['nuclear'], label='Nuclear', color='#9933ff', linewidth=2, marker='d')
plt.plot(energy_df['year'], energy_df['renewables'], label='Renewables', color='#00cc66', linewidth=2, marker='*', markersize=10)

plt.xlabel('Year', fontsize=12)
plt.ylabel('Energy Production (Exajoules)', fontsize=12)
plt.title('Global Primary Energy Production by Source', fontsize=14)

# Legend placement: 'best' lets matplotlib choose, or specify 'upper left', 'lower right', etc.
plt.legend(loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()
```

The legend is essential when showing multiple series. Without it, viewers cannot distinguish what each line represents. The `label` parameter in each `plot()` call provides the text that appears in the legend.

Notice the story the data tells: fossil fuels continue to dominate in absolute terms, but renewables show rapid growth especially in recent years. This is exactly the kind of pattern that visualisation makes clear.

## Figures with Multiple Subplots

Sometimes you want to show related but distinct plots together. Matplotlib's `subplot` function divides a figure into a grid of axes. The syntax `plt.subplot(rows, cols, index)` creates a subplot at a specific position, where index counts from 1 starting at the top-left and moving right then down.

```python
# Create a 2x2 grid of subplots
plt.figure(figsize=(12, 10))

# Top-left: Temperature anomaly
plt.subplot(2, 2, 1)
plt.plot(temp_df['year'], temp_df['anomaly'], color='crimson', linewidth=2)
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
plt.xlabel('Year')
plt.ylabel('Anomaly (°C)')
plt.title('Global Temperature Anomaly')

# Top-right: Fossil fuels
plt.subplot(2, 2, 2)
plt.plot(energy_df['year'], energy_df['coal'], label='Coal', color='#4a4a4a', linewidth=2)
plt.plot(energy_df['year'], energy_df['oil'], label='Oil', color='#2d5a27', linewidth=2)
plt.plot(energy_df['year'], energy_df['gas'], label='Gas', color='#1a75ff', linewidth=2)
plt.xlabel('Year')
plt.ylabel('Exajoules')
plt.title('Fossil Fuel Production')
plt.legend(fontsize=9)

# Bottom-left: Clean energy
plt.subplot(2, 2, 3)
plt.plot(energy_df['year'], energy_df['nuclear'], label='Nuclear', color='#9933ff', linewidth=2)
plt.plot(energy_df['year'], energy_df['renewables'], label='Renewables', color='#00cc66', linewidth=2)
plt.xlabel('Year')
plt.ylabel('Exajoules')
plt.title('Low-Carbon Energy Production')
plt.legend(fontsize=9)

# Bottom-right: Renewables growth (zoomed)
plt.subplot(2, 2, 4)
plt.plot(energy_df['year'], energy_df['renewables'], color='#00cc66', linewidth=2, marker='o')
plt.fill_between(energy_df['year'], energy_df['renewables'], alpha=0.3, color='#00cc66')
plt.xlabel('Year')
plt.ylabel('Exajoules')
plt.title('Renewable Energy Growth')

plt.tight_layout()  # Adjusts spacing to prevent overlap
plt.show()
```

The `fill_between()` function in the bottom-right subplot adds a shaded area under the line, which can emphasise cumulative growth or highlight the area of interest.

An alternative approach uses `plt.subplots()` (note the 's'), which returns both the figure and an array of axes objects. This object-oriented approach offers more control.

```python
# Object-oriented approach with subplots()
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))

# Left plot
axes[0].plot(temp_df['year'], temp_df['anomaly'], color='crimson', linewidth=2)
axes[0].axhline(y=0, color='gray', linestyle='--', alpha=0.5)
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Temperature Anomaly (°C)')
axes[0].set_title('Global Temperature Trend')

# Right plot
axes[1].plot(energy_df['year'], energy_df['renewables'], color='#00cc66', linewidth=2, marker='o')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Energy Production (EJ)')
axes[1].set_title('Renewable Energy Production')

# Add a main title for the entire figure
fig.suptitle('Climate Change and Energy Transition', fontsize=14, y=1.02)

plt.tight_layout()
plt.show()
```

## Histograms

Histograms show the distribution of a single variable by counting how many values fall into each bin (range). They answer questions like "How are these values spread out?" and "What's the most common range?"

Let's create some data representing household electricity consumption.

```python
# Simulated household monthly electricity consumption in kWh
# Using normal distribution with some realistic parameters
np.random.seed(42)  # For reproducibility

household_consumption = np.random.normal(loc=350, scale=100, size=500)
# Clip to avoid unrealistic negative values
household_consumption = np.clip(household_consumption, 50, 800)

print(f"Sample statistics:")
print(f"  Mean: {np.mean(household_consumption):.1f} kWh")
print(f"  Median: {np.median(household_consumption):.1f} kWh")
print(f"  Std Dev: {np.std(household_consumption):.1f} kWh")
```

```python
# Basic histogram
plt.figure(figsize=(10, 6))

plt.hist(household_consumption, bins=20, color='steelblue', edgecolor='white')

plt.xlabel('Monthly Electricity Consumption (kWh)', fontsize=12)
plt.ylabel('Number of Households', fontsize=12)
plt.title('Distribution of Household Electricity Consumption', fontsize=14)

plt.show()
```

The `bins` parameter controls how many bars the histogram has. Fewer bins give a smoother but less detailed view; more bins show finer detail but can be noisy. There's no single correct choice, it depends on your data and what you're trying to show.

The `edgecolor='white'` separates the bars visually, making them easier to distinguish.

```python
# Comparing different bin counts
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, bins in zip(axes, [10, 25, 50]):
    ax.hist(household_consumption, bins=bins, color='steelblue', edgecolor='white')
    ax.set_xlabel('Consumption (kWh)')
    ax.set_ylabel('Count')
    ax.set_title(f'{bins} bins')

plt.tight_layout()
plt.show()
```

You can add reference lines to provide context. For instance, showing the mean or median helps viewers understand where the centre of the distribution lies.

```python
plt.figure(figsize=(10, 6))

plt.hist(household_consumption, bins=25, color='steelblue', edgecolor='white', alpha=0.7)

# Add vertical lines for mean and median
mean_val = np.mean(household_consumption)
median_val = np.median(household_consumption)

plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.0f} kWh')
plt.axvline(median_val, color='orange', linestyle='-', linewidth=2, label=f'Median: {median_val:.0f} kWh')

plt.xlabel('Monthly Electricity Consumption (kWh)', fontsize=12)
plt.ylabel('Number of Households', fontsize=12)
plt.title('Distribution of Household Electricity Consumption', fontsize=14)
plt.legend()

plt.show()
```

## Comparing Distributions: Multiple Histograms

Let's create data for two different groups and compare their distributions.

```python
# Household consumption: with and without solar panels
np.random.seed(42)

without_solar = np.random.normal(loc=380, scale=90, size=400)
without_solar = np.clip(without_solar, 100, 700)

with_solar = np.random.normal(loc=220, scale=80, size=300)
with_solar = np.clip(with_solar, 50, 500)
```

```python
# Overlapping histograms with transparency
plt.figure(figsize=(10, 6))

plt.hist(without_solar, bins=20, alpha=0.6, color='coral', edgecolor='white', label='Without Solar')
plt.hist(with_solar, bins=20, alpha=0.6, color='teal', edgecolor='white', label='With Solar')

plt.xlabel('Monthly Grid Electricity (kWh)', fontsize=12)
plt.ylabel('Number of Households', fontsize=12)
plt.title('Grid Electricity Consumption: Solar vs Non-Solar Households', fontsize=14)
plt.legend()

plt.show()
```

The alpha (transparency) parameter is essential when overlapping histograms, otherwise one would hide the other. The comparison clearly shows that households with solar panels draw less from the grid on average.

## Stacked Histograms

Stacked histograms show distributions for multiple groups while also showing the total. They work best when you want to see both the individual contributions and the combined whole.

```python
# Create data for three building types
np.random.seed(42)

apartments = np.random.normal(loc=250, scale=60, size=200)
houses = np.random.normal(loc=400, scale=100, size=300)
commercial = np.random.normal(loc=600, scale=150, size=150)

# Clip to reasonable ranges
apartments = np.clip(apartments, 100, 450)
houses = np.clip(houses, 150, 750)
commercial = np.clip(commercial, 250, 1000)
```

```python
plt.figure(figsize=(10, 6))

# Pass a list of arrays to create a stacked histogram
plt.hist([apartments, houses, commercial], 
         bins=20, 
         stacked=True,
         color=['#3498db', '#e74c3c', '#2ecc71'],
         edgecolor='white',
         label=['Apartments', 'Houses', 'Commercial'])

plt.xlabel('Monthly Electricity Consumption (kWh)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.title('Electricity Consumption by Building Type', fontsize=14)
plt.legend()

plt.show()
```

The stacked histogram shows that apartments cluster at lower consumption, houses form the middle range with the most variability, and commercial buildings have the highest and most variable consumption. The total height of each bar shows how many buildings of all types fall in that consumption range.

## Side-by-Side Histograms

An alternative to stacking is placing histograms side by side, which can make individual distributions easier to compare.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

axes[0].hist(apartments, bins=15, color='#3498db', edgecolor='white')
axes[0].set_xlabel('kWh')
axes[0].set_ylabel('Count')
axes[0].set_title('Apartments')
axes[0].axvline(np.mean(apartments), color='black', linestyle='--', label=f'Mean: {np.mean(apartments):.0f}')
axes[0].legend()

axes[1].hist(houses, bins=15, color='#e74c3c', edgecolor='white')
axes[1].set_xlabel('kWh')
axes[1].set_title('Houses')
axes[1].axvline(np.mean(houses), color='black', linestyle='--', label=f'Mean: {np.mean(houses):.0f}')
axes[1].legend()

axes[2].hist(commercial, bins=15, color='#2ecc71', edgecolor='white')
axes[2].set_xlabel('kWh')
axes[2].set_title('Commercial')
axes[2].axvline(np.mean(commercial), color='black', linestyle='--', label=f'Mean: {np.mean(commercial):.0f}')
axes[2].legend()

fig.suptitle('Electricity Consumption Distribution by Building Type', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

The `sharey=True` parameter ensures all subplots use the same y-axis scale, making visual comparisons valid. Without this, matplotlib would scale each subplot independently, which could be misleading.

## Bar Charts

Bar charts display categorical data, showing how a measured value varies across different groups or categories. While histograms show the distribution of a continuous variable, bar charts compare discrete categories.

```python
# CO2 emissions by country (approximate 2023 figures in million tonnes)
countries = ['China', 'USA', 'India', 'Russia', 'Japan', 'Germany', 'Iran', 'S. Korea']
emissions = [11680, 4930, 2830, 1760, 1030, 640, 690, 620]

plt.figure(figsize=(10, 6))

plt.bar(countries, emissions, color='#d35400', edgecolor='white')

plt.xlabel('Country', fontsize=12)
plt.ylabel('CO₂ Emissions (Million Tonnes)', fontsize=12)
plt.title('Top CO₂ Emitting Countries (2023)', fontsize=14)

# Rotate labels for readability
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Horizontal bar chart (useful when labels are long)
plt.figure(figsize=(10, 6))

# Sort by value for better visual hierarchy
sorted_indices = np.argsort(emissions)
sorted_countries = [countries[i] for i in sorted_indices]
sorted_emissions = [emissions[i] for i in sorted_indices]

plt.barh(sorted_countries, sorted_emissions, color='#3498db', edgecolor='white')

plt.xlabel('CO₂ Emissions (Million Tonnes)', fontsize=12)
plt.title('CO₂ Emissions by Country (2023)', fontsize=14)

plt.tight_layout()
plt.show()
```

## Saving Figures

Once you've created a figure you're happy with, you can save it to a file. Matplotlib supports various formats including PNG, PDF, SVG, and EPS.

```python
# Create a figure
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(temp_df['year'], temp_df['anomaly'], color='crimson', linewidth=2)
ax.axhline(y=0, color='gray', linestyle='--', alpha=0.5)
ax.set_xlabel('Year')
ax.set_ylabel('Temperature Anomaly (°C)')
ax.set_title('Global Temperature Anomaly')

# Save to file
# The dpi parameter controls resolution (dots per inch)
fig.savefig('temperature_anomaly.png', dpi=150, bbox_inches='tight')

print("Figure saved as temperature_anomaly.png")
plt.show()
```

The `bbox_inches='tight'` parameter ensures that labels and titles aren't cut off in the saved image. Without it, elements near the edges sometimes get cropped.
