## Setup Code (Run This First)

Run this cell to set up your database and import necessary libraries.

```python
# Import libraries
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Create database connection
conn = sqlite3.connect('mystical_academy.db')
cursor = conn.cursor()

# Create tables
cursor.execute('''
CREATE TABLE IF NOT EXISTS wizards (
    wizard_id INTEGER PRIMARY KEY,
    wizard_name TEXT NOT NULL,
    house TEXT NOT NULL,
    year_of_study INTEGER NOT NULL,
    magical_power INTEGER NOT NULL
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS spells (
    spell_id INTEGER PRIMARY KEY,
    spell_name TEXT NOT NULL,
    spell_type TEXT NOT NULL,
    difficulty_level INTEGER NOT NULL,
    mana_cost INTEGER NOT NULL
)
''')

cursor.execute('''
CREATE TABLE IF NOT EXISTS castings (
    casting_id INTEGER PRIMARY KEY,
    wizard_id INTEGER NOT NULL,
    spell_id INTEGER NOT NULL,
    success BOOLEAN NOT NULL,
    casting_date DATE NOT NULL,
    FOREIGN KEY (wizard_id) REFERENCES wizards(wizard_id),
    FOREIGN KEY (spell_id) REFERENCES spells(spell_id)
)
''')

# Insert sample data
wizards_data = [
    (1, 'Zara Moonwhisper', 'Phoenix Tower', 3, 85),
    (2, 'Felix Starweaver', 'Dragon Hall', 2, 72),
    (3, 'Luna Shadowmend', 'Phoenix Tower', 4, 91),
    (4, 'Rowan Flameheart', 'Griffin Keep', 1, 68),
    (5, 'Sage Crystalwind', 'Dragon Hall', 3, 88),
    (6, 'Nova Dreamwalker', 'Griffin Keep', 2, 79)
]

spells_data = [
    (101, 'Levitation Charm', 'Movement', 2, 15),
    (102, 'Fireball Blast', 'Elemental', 4, 35),
    (103, 'Healing Touch', 'Restoration', 3, 25),
    (104, 'Shield Ward', 'Protection', 2, 20),
    (105, 'Lightning Strike', 'Elemental', 5, 45),
    (106, 'Invisibility Cloak', 'Illusion', 4, 30)
]

castings_data = [
    (1, 1, 101, True, '2024-01-15'),
    (2, 1, 103, True, '2024-01-20'),
    (3, 2, 101, False, '2024-01-18'),
    (4, 2, 104, True, '2024-01-22'),
    (5, 3, 102, True, '2024-01-16'),
    (6, 3, 105, True, '2024-01-25'),
    (7, 4, 101, True, '2024-01-17'),
    (8, 4, 104, False, '2024-01-21'),
    (9, 5, 103, True, '2024-01-19'),
    (10, 5, 106, True, '2024-01-24'),
    (11, 6, 102, False, '2024-01-20'),
    (12, 6, 104, True, '2024-01-23')
]

cursor.executemany('INSERT OR IGNORE INTO wizards VALUES (?,?,?,?,?)', wizards_data)
cursor.executemany('INSERT OR IGNORE INTO spells VALUES (?,?,?,?,?)', spells_data)
cursor.executemany('INSERT OR IGNORE INTO castings VALUES (?,?,?,?,?)', castings_data)

conn.commit()
print("✨ Database setup complete! The Mystical Academy database is ready.")
```
